```python
# import python libraries
import pandas as pd
import hvplot.pandas
import holoviews as hv
import osmnx as osm
```

```python
bu_boundary = osm.geocode_to_gdf(
                         "R10366327",
                         by_osmid=True
)

bu_boundary
```

```python
bu_boundary.plot()
```

```python
# make interactive map
bu_map = bu_boundary.hvplot(geo=True,  tiles="EsriImagery")
bu_map
```

```python
# Plot Cu boundary
bu_map = bu_boundary.reset_index().hvplot(
    # Add a basemap
    geo=True, tiles='EsriImagery',
    # Change the colors
    fill_color='white', fill_alpha=0.2,
    line_color='skyblue', line_width=5,
    # Change the image size
    frame_width=400, frame_height=400)

# Save the map as a file to put on the web
hv.save(bu_map, 'bu_map.html')

# Display the map
bu_map
```
